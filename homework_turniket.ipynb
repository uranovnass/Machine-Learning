```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Исходные данные
TP = 16
FN = 4
FP = 11
TN = 219

# Общее количество объектов
total = TP + FN + FP + TN

# 4. Accuracy модели, которая всегда отвечает "нет"
accuracy_always_no = (TN + 0) / total

# Accuracy нашей модели
accuracy_model = (TP + TN) / total

print("=== 4. Accuracy ===")
print(f"Всего объектов: {total}")
print(f"Accuracy модели «всегда нет»: {accuracy_always_no:.3f} ({accuracy_always_no*100:.1f}%)")
print(f"Accuracy нашей модели: {accuracy_model:.3f} ({accuracy_model*100:.1f}%)")


# 5. Какая ошибка дороже?
print("\n=== 5. Ошибка ===")
print("Для турникета FN дороже, потому что чужой человек может быть ошибочно принят за своего.")
print("Главная метрика: Recall.")


# 6. Сдвиг порога
# 3 FN -> TP
# 5 TN -> FP

new_TP = TP + 3
new_FN = FN - 3
new_FP = FP + 5
new_TN = TN - 5

# Новые метрики
new_precision = new_TP / (new_TP + new_FP)
new_recall = new_TP / (new_TP + new_FN)
new_f1 = 2 * new_precision * new_recall / (new_precision + new_recall)

print("\n=== 6. После снижения порога ===")

print("\nНовая матрица ошибок:")
print(f"TP = {new_TP}")
print(f"FN = {new_FN}")
print(f"FP = {new_FP}")
print(f"TN = {new_TN}")

print("\nPrecision:", round(new_precision, 3))
print("Recall:", round(new_recall, 3))
print("F1:", round(new_f1, 3))
```
