```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="I0", bins=20, kde=True)

plt.title("Распределение I0")
plt.xlabel("I0")
plt.ylabel("Количество наблюдений")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Class")

plt.title("Распределение классов ткани")
plt.xlabel("Класс")
plt.ylabel("Количество образцов")
plt.xticks(rotation=45)
plt.show()
```

```python
import pandas as pd

df = pd.read_csv("../data/data.csv")

print(df.shape)
```

```python
print("Количество объектов каждого класса:")
print(df["Class"].value_counts())

print("\nДоли классов (%):")
print((df["Class"].value_counts(normalize=True) * 100).round(2))

print("\nСредние значения числовых признаков по классам:")
print(df.groupby("Class").mean(numeric_only=True).round(2))
```

### Наблюдения

1. В датасете представлены 6 классов тканей. Наиболее многочисленным является класс `adi` — 22 образца (20,75%), а наименее многочисленным — `con` — 14 образцов (13,21%). При этом распределение классов относительно равномерное: ни один класс не содержит даже четверти всех наблюдений.

2. Числовые признаки заметно различаются между классами. Например, среднее значение `I0` для класса `adi` составляет 2052,05, тогда как для `gla` — 238,32. Это показывает, что значения электрических характеристик существенно зависят от класса ткани.

3. Класс `adi` отличается высокими средними значениями сразу нескольких признаков: `I0` = 2052,05, `Area` = 24889,74, `Max.IP` = 194,60, `DR` = 324,71 и `P` = 2138,75. В сравнении с большинством остальных классов эти показатели значительно выше, поэтому данные признаки могут быть полезны при изучении различий между классами.
